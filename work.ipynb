**반복문 while**  
while n > 0 = "n > 0 때까지 반복해"

```python
n = 5
while n > 0:
    print(n, "초 전")
    n = n - 1
print("발사!")
```

**문자열 - 포매팅**   
문자열 안에 또 다른 문자열을 삽입하기  
숫자를 넣기 위해서는 %d, 문자열을 넣기 위해서는 %s를 씀

```python
"현재 습도는 %d%%입니다."  %17 , "%d일의 습도는 %d%%입니다." %(12, 17) , "현재 습도는 %s도입니다." %"십칠"  
# 습도의 단위인 퍼센트 기호를 문자열에 포함하기 위해서는 %%로 두번 입력하면 됨
```

%f 사이에 적은 소수점만큼 소수점 자리수가 정리됨

```python
"원주율은 보통 %f로 씁니다." %3.1415926535897932 , "원주율은 보통 %0.2f로 씁니다." %3.1415926535897932  
#소수점 포인트 앞의 숫자는 문자열의 전체 길이를 의미함  
# %0.4f는 0을 생략하여 %.4f처럼 사용하기도 한다
```

**딕셔너리**  
Key와 Value를 한 쌍으로 가지는 자료형  
Key부분에는 변하지 않는 값만 넣을 수 있음. Value에는 아무 값이나 넣을 수 있음.

```python
정보 = {"이름":"김원경", "나이":36, "사는곳":"경기광주"}  
정보["사는곳"]
```

```python
정보 = {"이름":"김원경", "나이":36, "사는곳":"경기광주"}  
for 정보 in 정보.keys():  
    print(정보)
```

```python
정보 = {"이름":"김원경", "나이":36, "사는곳":"경기광주"}  
for 정보 in 정보.values():  
    print(정보)
```
